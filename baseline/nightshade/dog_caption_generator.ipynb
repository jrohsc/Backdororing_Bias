```python
import pandas as pd
from transformers import GPT2Tokenizer
```

```python
# Initialize the GPT-2 tokenizer
tokenizer = GPT2Tokenizer.from_pretrained('gpt2')
```

```python
df = pd.read_pickle('../../../midjourney datasets/latest update/final_dataset.pkl')
len(df)
```

```python
# Specify the exact word you're looking for
specific_word = r'\bdog\b'  # The \b denotes a word boundary

# Using regex=True to enable regular expression functionality
filtered_df = df[df['Prompt'].str.contains(specific_word, case=False, na=False, regex=True)]
len(filtered_df)
```

```python
filtered_df.head(2)
```

```python
# Filter prompts based on token length
final_filtered_df = filtered_df[filtered_df['Prompt'].apply(lambda x: len(tokenizer.encode(x)) <= 20)]
len(final_filtered_df)
```

```python
# Drop duplicates based on the 'Prompt' column
final_filtered_df = final_filtered_df.drop_duplicates(subset=['Prompt'])
# Reset the index of the DataFrame
final_filtered_df = final_filtered_df.reset_index(drop=True)
len(final_filtered_df)
```
