```python
import os
import pickle
import numpy as np
from PIL import Image
from tqdm import tqdm
import pandas as pd

```

```python
# Paths
prompt_path = "dog_prompts.pkl"  # the pickle file containing the "Prompt" column
image_folder = "dalle images/dog"       # folder with images named like 0.png, 1.png, ...
output_dir = "data/dog"
os.makedirs(output_dir, exist_ok=True)

```

```python
# Load prompt DataFrame
with open(prompt_path, "rb") as f:
    df = pickle.load(f)

```

```python
# Iterate through the DataFrame with tqdm
for idx, row in tqdm(df.iterrows(), total=len(df), desc="Converting to .p"):
    prompt = row["Prompt"]
    image_path = os.path.join(image_folder, f"{idx}.png")

    try:
        img = Image.open(image_path).convert("RGB")
        img_np = np.array(img)
    except Exception as e:
        print(f"❌ Skipping {idx}: error loading {image_path} — {e}")
        continue

    data = {
        "img": img_np,
        "text": prompt
    }

    with open(os.path.join(output_dir, f"{idx}.p"), "wb") as f:
        pickle.dump(data, f)

print(f"✅ Saved {len(os.listdir(output_dir))} .p files to: {output_dir}")
```
